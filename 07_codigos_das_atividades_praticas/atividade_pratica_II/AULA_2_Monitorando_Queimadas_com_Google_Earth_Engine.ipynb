`Conhecendo o GEE`:

O [Google Earth Engine (GEE)](https://earthengine.google.com/) é uma plataforma na nuvem poderosa para análise geoespacial em larga escala. O GEE combina um catálogo de vários petabytes de imagens de satélite e conjuntos de dados geoespaciais com capacidades de análises em escala planetária. Cientistas, pesquisadores e desenvolvedores usam o Earth Engine para detectar mudanças, mapear tendências e quantificar diferenças na superfície da Terra. O Earth Engine está disponível para uso comercial e permanece gratuito para uso acadêmico e de pesquisa. Assista o [vídeo](https://www.youtube.com/watch?v=gKGOeTFHnKY) e veja exemplos de aplicações do GEE. A grande vantagem do GEE é que realizamos o carregamento e processamento dos dados na nuvem, sobrando tempo assim para analisarmos os resultados de um trabalho que estamos realizando.

A [base de dados](https://developers.google.com/earth-engine/datasets?hl=pt-br) do GEE é extensa e abrange dados de tempo, clima, superfície terrestre e oceânica. Para meteorologia os principais dados disponíveis são:

1. [Imagens GOES-19](https://developers.google.com/earth-engine/datasets/tags/goes-19)
2. [Precipitação](https://developers.google.com/earth-engine/datasets/tags/precipitation)
3. [ERA5](https://developers.google.com/earth-engine/datasets/catalog/ECMWF_ERA5_LAND_HOURLY)

5. [NDVI](https://developers.google.com/earth-engine/datasets/tags/ndvi)

Quando trabalhamos com mapas, muita das vezes estamos interessados numa região específica. No nosso caso, a nossa região de interesse é o Brasil e gostaríamos de dar um zoom no território brasileiro. Então podemos centralizar o nosso mapa no Brasil, e para isto vamos inserir os argumentos **center** e **zoom** na função `geemap.Map()`.
- `center`: Recebe uma lista com [latitude, longitude]

- `zoom`: Nível de zoom (geralmente entre 1-20, onde valores maiores significam zoom mais próximo)

```python
# cria o mapa com centro nas coordenadas [latitude, longitude] e zoom inicial
Map = geemap.Map(center=[-14, -55], zoom=4)

# adicionando basemap
Map.add_basemap('SATELLITE')

# exibe na tela
Map
```

Para detalhar melhor uma região geralmente plotamos o contorno daquela região no mapa. Assim o GEE tem uma base de dados dos contornos dos municípios para o mundo todo. Esta base de dados é uma Feature Collection de [nível 2](https://developers.google.com/earth-engine/datasets/catalog/FAO_GAUL_2015_level2).

Primeiro selecionamos a base de dados global de municípios com a função `ee.FeatureCollection('FAO/GAUL/2015/level2')` e depois escolhemos o município de **Ribeirão Preto** aplicando a função `filter(ee.Filter.eq('ADM2_NAME', 'Ribeirao Preto'))`.
> **LEMBRE-SE**: Apenas a primeira letra de cada palavra dever ser maiúscula e todas as letras devem estar sem acentos.  

```python
# shapefile do município de Ribeirão Preto
ribeirao_preto = ee.FeatureCollection('FAO/GAUL/2015/level2').filter(ee.Filter.eq('ADM2_NAME', 'Ribeirao Preto'))
```

```python
# mostrando a "FeatureCollection" que contêm o contorno (ou seja, o shapefile) do município de Ribeirão Preto
ribeirao_preto
```

Iremos utilizar a função `centerObject` para centralizar o nosso mapa em Ribeirão Preto. Como argumento iremos inserir a *FeatureCollection* do município de Ribeirão Preto e um zoom de valor 11. Além disso, iremos utilizar a função `addLayer` para plotar o contorno de Ribeirão Preto no mapa.

```python
# criando um mapa interativo
Map = geemap.Map()

# centra o mapa na região
Map.centerObject(ribeirao_preto, zoom=11)

# adicionando basemap
Map.add_basemap('SATELLITE')

# contorno do município de Ribeirão Preto
style = {'color': 'yellow', 'fillColor': '00000000'}
Map.addLayer(ribeirao_preto.style(**style), {}, 'Ribeirão Preto')

# exibe na tela
Map
```

Agora se você precisar analisar o mapa a nível estadual, podemos plotar o contorno do Estado de São Paulo (SP). Para isto, utilizaremos a base de dados dos contornos dos Estados para o mundo todo. Esta base de dados é uma Feature Collection de [nível 1](https://developers.google.com/earth-engine/datasets/catalog/FAO_GAUL_2015_level1).

Primeiro selecionamos a base de dados global de Estados com a função `ee.FeatureCollection('FAO/GAUL/2015/level1')` e depois escolhemos o Estado de São Paulo aplicando a função `filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo'))`.

```python
# shapefile do Estado de São Paulo
estado_sp = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo'))
```

Iremos também fazer alterações no estilo do contorno do Estado, mudando a largura e tipo da linha do contorno.

```python
# criando um mapa interativo
Map = geemap.Map()

# centra o mapa na região
Map.centerObject(estado_sp, zoom=7)

# adicionando basemap
Map.add_basemap('SATELLITE')

# contorno de São Paulo
style = {
         'color': 'red',
         'fillColor': '00000000',  # transparente
         'width': 3,               # espessura da linha (em pixels)
         'lineType': 'dotted',     # tipo de linha. Opções: 'solid', 'dashed', 'dotted'
         }
Map.addLayer(estado_sp.style(**style), {}, 'Estado de SP')

# exibe na tela
Map
```

# **3° Passo:** Carregando Dados no Google Earth Engine (GEE)

`Objetos e Métodos do GEE`:

Para acessar os dados do GEE, realizar cálculos e gerar gráficos nativos utilizamos os `Objetos` e `Métodos` próprios do GEE. São maneiras de acessar uma imagem, coleção de imagens, geometrias e realizar reduções (cálculos espaciais). Os [Objetos](https://developers.google.com/earth-engine/guides/objects_methods_overview) mais utilizados são:

1. [`Imagem`](https://developers.google.com/earth-engine/guides/image_overview): dados raster no GEE são representados como imagens. Imagens são compostas de uma ou mais bandas (canais) e possuem metadados com grupos de propriedades.  

2. [`ImagemCollection`](https://developers.google.com/earth-engine/guides/ic_creating): é um conjunto ou sequência de imagens e o acesso é através de um *asset ID*.

3. [`Geometry`](https://developers.google.com/earth-engine/guides/geometries): dados vetoriais.

4. [`Feature`](https://developers.google.com/earth-engine/guides/features): geometria com atributos.

5. [`FeatureCollection`](https://developers.google.com/earth-engine/guides/feature_collections): grupo ou coleção de geometrias. É uma coleção de features com propriedades e atributos.

6. [`Reducer`](https://developers.google.com/earth-engine/guides/reducers_intro): calcula estatísticas e agregações espaciais/temporais.

7. [`Join`](https://developers.google.com/earth-engine/guides/joins_intro): combina conjunto de dados baseado no tempo, localização ou atributo.

8. [`Array`](https://developers.google.com/earth-engine/guides/arrays_intro): trabalha com análises multidimensionais.

9. [`Chart`](https://developers.google.com/earth-engine/guides/charts_overview): utilizado para gerar gráficos.

Como exemplo, iremos carregar as coleção de imagens de [reflectância](https://developers.google.com/earth-engine/datasets/catalog/MODIS_061_MYD09GA?hl=pt-br#bands) de superfície do sensor MODIS do satélite AQUA. Esta base de dados fornece informações da reflectância a nível de superfície em escala global com frequência temporal diária para o período de julho de 2002 até o momento presente.

O código Snippet dessa base de dados é `MODIS/061/MYD09GA`. Assim, iremos utilizar o objeto `ee.ImageCollection` para carregar a coleção de imagens entre 1 e 31 de agosto de 2024. Segue abaixo uma explicação de cada uma das funções que utilizaremos:


---

1. Carregando a Coleção de Imagens: `modis = ee.ImageCollection('MODIS/061/MYD09GA')`

    - ee.ImageCollection(): Função que acessa uma coleção de imagens no GEE

    - 'MODIS/061/MYD09GA': Caminho do dataset específico:

        - MODIS: Sensor MODIS

        - 061: Versão 6.1 dos dados

        - MYD09GA: Produto de refletância de superfície global

2. Filtragem Temporal: `.filterDate('2024-08-01','2024-09-01')`

    - filterDate(): Filtra imagens entre 1 e 31 de agosto de 2024

    - Argumentos:

        - '2024-08-01': Data inicial (formato ISO: YYYY-MM-DD)

        - '2024-09-31': Data final (exclusiva)

3. Processamento das Imagens: `.map(lambda image: image.multiply(0.0001).copyProperties(image, ['system:time_start']))`
    - map(): Aplica uma função a cada imagem da coleção

    - Função Lambda:

        - image.multiply(0.0001): Multiplica os valores dos pixels por 0,0001 (os dados MODIS originais são escalonados ×10,000)

        - copyProperties(image, ['system:time_start']): Preserva os metadados originais, especificamente a propriedade de timestamp


---

A primeira base de dados relacionada a queimadas que iremos conhecer é a [Fire Information for Resource Management System(FIRMS)](https://developers.google.com/earth-engine/datasets/catalog/FIRMS).
```python
# região de interesse
estado_sp = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo'))

# carrega os dados
firms = ee.ImageCollection('FIRMS') \
          .filterBounds(estado_sp) \
          .filterDate('2024-08-01','2024-09-01') \
          .select('T21') \
          .map(lambda img: img.subtract(273.15).copyProperties(img, img.propertyNames()))

# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(estado_sp, 7)

# parâmetros de visualização
param_vis = {'min': 30, 'max': 150, 'palette': ['red', 'orange', 'yellow']}

# adicionando basemap
Map.add_basemap('SATELLITE')

# plotando mapa
Map.addLayer(firms.mean().clip(estado_sp), param_vis, 'Temp. em agosto/2024')

# contorno do estado de SP
style = {'color': 'yellow', 'fillColor': '00000000'}
Map.addLayer(estado_sp.style(**style), {}, 'SP')

# barra de cores
Map.add_colorbar(param_vis, label='Temperatura do Pixel (°C)', orientation='horizontal')

# exibe na tela
Map
```

#**PARTE 2)**: Detecção de Focos de Calor com os Dados [VIIRS (NOAA-20)](https://developers.google.com/earth-engine/datasets/catalog/NASA_LANCE_NOAA20_VIIRS_C2?hl=pt-br)

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/viirs.png?raw=true)

O produto de detecção de incêndios ativos do sensor Visible Infrared Imaging Radiometer Suite (VIIRS) do satélite [NOAA-20](https://space.oscar.wmo.int/satellites/view/noaa_20), denominado [VIIRS (NOAA-20) Band 375m Active Fire](https://developers.google.com/earth-engine/datasets/catalog/NASA_LANCE_NOAA20_VIIRS_C2?hl=pt-br#description) baseia-se nos dados de resolução nominal de 375 m do instrumento. Em comparação com outros produtos de detecção de incêndios por satélite de resolução mais baixa (≥ 1 km), os dados de 375 m oferecem uma resposta maior a incêndios em áreas relativamente pequenas, além de um mapeamento melhor de grandes perímetros de incêndio. Consequentemente, os dados são adequados para uso no suporte ao gerenciamento de incêndios (por exemplo, sistemas de alerta quase em tempo real), bem como em outras aplicações científicas que exigem melhor fidelidade no mapeamento de incêndios. Além da vantagem em relação a resolução espacial (375 m), são disponibilizados também os dados da Potência Radiativa de Incêndio (FRP), que descreve a potência radiativa de incêndio integrada ao pixel em megawatts (MW). O período de dados compreende desde outubro de 2023 até o momento presente.

No exemplo a seguir iremos plotar os dados da temperatura de brilho do canal 4 (infravermelho médio 3,55 a 3,93 μm) e FRP para o dia 23 de agosto de 2024.





```python
# região de interesse
estado_sp = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo'))

# carrega os dados da temperatura do canal 4
ti4_viirs_noaa20 = ee.ImageCollection('NASA/LANCE/NOAA20_VIIRS/C2') \
                     .filterBounds(estado_sp) \
                     .filterDate('2024-08-23','2024-08-24') \
                     .select('Bright_ti4') \
                     .map(lambda img: img.subtract(273.15).copyProperties(img, img.propertyNames()))

# carrega os dados de frp
frp_viirs_noaa20 = ee.ImageCollection('NASA/LANCE/NOAA20_VIIRS/C2') \
                     .filterBounds(estado_sp) \
                     .filterDate('2024-08-23','2024-08-24') \
                     .select('frp')

# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(estado_sp, 7)

# parâmetros de visualização
param_vis_ti4 = {'min': 30, 'max': 150, 'palette': ['red', 'orange', 'yellow']}
param_vis_frp = {'min': 0, 'max': 100, 'palette': ['red', 'orange', 'yellow']}

# adicionando basemap
Map.add_basemap('SATELLITE')

# mapa de temperatura do canal 4
Map.addLayer(ti4_viirs_noaa20.mean().clip(estado_sp), param_vis_ti4, 'ti4 (°C) - 23/ago/2024')

# mapa de FRP
Map.addLayer(frp_viirs_noaa20.mean().clip(estado_sp), param_vis_frp, 'frp (MW) - 23/ago/2024')

# contorno do estado de SP
style = {'color': 'yellow', 'fillColor': '00000000'}
Map.addLayer(estado_sp.style(**style), {}, 'SP')

# barra de cores
Map.add_colorbar(param_vis_ti4, label='Temperatura do Pixel (°C)', orientation='horizontal')
Map.add_colorbar(param_vis_frp, label='FRP (MW)', orientation='horizontal')

# exibe na tela
Map
```

#**PARTE 3)**: Detecção de Focos de Calor com os Dados [GOES-16](https://developers.google.com/earth-engine/datasets/catalog/NOAA_GOES_16_FDCF?hl=pt-br#bands)

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/goes16.png?raw=true)

Para dados de focos de calor podemos utilizar as informações do satélite geoestacionário [GOES-16](https://space.oscar.wmo.int/satellites/view/goes_16). A grande vantagem é que temos imagens a cada 10min fornecendo dados de temperatura, área do fogo, FRP e categorias de máscara de fogo. Esta base de dados é denominada [GOES-16 FDCF Series ABI Level 2 Fire/Hot Spot Characterization Full Disk](https://developers.google.com/earth-engine/datasets/catalog/NOAA_GOES_16_FDCF?hl=pt-br#bands), possui resolução espacial de 2 km e esta disponível de 24/05/2017 à 07/04/2025. Para a nossa data de estudo (23/agosto/2024) estava em operação o satélite GOES-16, se você precisar de dados mais atuais utilize os [dados](https://developers.google.com/earth-engine/datasets/catalog/NOAA_GOES_19_FDCF#bands) do satélite GOES-19.

As seguintes bandas são disponíveis nestes dados:

1. `Area`: área do fogo em metros quadrados.
2. `Temp`: temperatura do fogo em Kelvin. Temos um fator de escala de *0.0549367* e offset de *400*.
3. `Mask`: categorias da máscara de incêndio. Os valores dos pixels na imagem da máscara de incêndio identificam uma categoria de incêndio e informações de diagnóstico associadas à execução do algoritmo. As seis categorias de incêndio incluem: pixel de incêndio de boa qualidade; pixel de incêndio saturado; pixel de incêndio contaminado por nuvens; pixel de incêndio de alta probabilidade; pixel de incêndio de média probabilidade e pixel de incêndio de baixa probabilidade.

4. `Power`: Fire radiative power (FRP) em MW.

No exemplo a seguir selecionaremos os focos de calor do dia 23/agosto/2024 utilizando apenas aqueles com categorias de mascara de fogo entre `10 e 35`. Note que a variável `fire_mask` armazena a contagem de detecções de fogo para cada pixel ao longo do período analisado. O valor que você vê nela é o resultado da soma da condição True para cada imagem na coleção. Assim, para cada imagem, o código verifica o valor da banda `'Mask'` e retorna `True (1)` ou `False (0)` dependendo se ele está no intervalo de 10 a 35. A função `.sum()` no final soma esses valores True ao longo de todas as imagens. Portanto, o valor de um pixel em fire_mask pode variar de:
 - `0`: O pixel nunca foi classificado como fogo (valores 10-35) em nenhuma das imagens analisadas.
 - `Até N`: O pixel foi classificado como fogo em todas as N imagens da sua coleção (onde N é o número total de imagens no período). Este valor representa a frequência de detecção, não a intensidade do fogo.



```python
#=================================================================================================#
#                                  CARREGA OS DADOS
#=================================================================================================#
# região de interesse
estado_sp = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo'))

# carrega os dados GOES-16
goes16 = ee.ImageCollection('NOAA/GOES/16/FDCF') \
           .filterBounds(estado_sp) \
           .filterDate('2024-08-23','2024-08-24') \
           .filter(ee.Filter.calendarRange(0, 50, 'minute')) # seleciona apenas imagens capturadas a cada 10min. Se for a cada a cada 15min (0, 45, 'minute'), a cada hora seria (0, 0, 'minute')

#=================================================================================================#
#                                 APLICA FATOR DE ESCALA E OFFSET
#=================================================================================================#
# temperatura (Temp) em Celsius com aplicação de fator de escala e offset
temp_goes16 = goes16.map(lambda img: img.select('Temp')
                                        .multiply(0.0549367)
                                        .add(400)
                                        .subtract(273.15)
                                        .copyProperties(img, img.propertyNames())).mean()

# máscara de fogo ativo (valores 10-35)
fire_mask = goes16.map(lambda img: img.select('Mask').gte(10).And(img.select('Mask').lte(35))).sum()

# FRP (Fire Radiative Power)
frp_goes16 = goes16.map(lambda img: img.select('Power')).mean()

#=================================================================================================#
#                                    PLOTA FIGURA
#=================================================================================================#
# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(estado_sp, 7)

# parâmetros de visualização
param_vis_temp = {'min': 30, 'max': 150, 'palette': ['blue', 'cyan', 'green', 'yellow', 'orange', 'red']}
param_vis_mask = {'min': 0, 'max': 1, 'palette': ['skyblue']}
param_vis_frp = {'min': 0, 'max': 1000, 'palette': ['yellow', 'orange', 'red']}

# adicionando basemap
Map.add_basemap('SATELLITE')

# mapa de temperatura
Map.addLayer(temp_goes16.clip(estado_sp), param_vis_temp, 'GOES-16 Temp (°C) - 23/ago/2024')

# mapa de máscara de incêndio com selfMask
Map.addLayer(fire_mask.selfMask().clip(estado_sp), param_vis_mask, 'GOES-16 Fire Mask - 23/ago/2024')

# mapa de FRP
Map.addLayer(frp_goes16.selfMask().clip(estado_sp), param_vis_frp, 'GOES-16 FRP (MW) - 23/ago/2024')

# contorno do estado de SP
style = {'color': 'yellow', 'fillColor': '00000000'}
Map.addLayer(estado_sp.style(**style), {}, 'SP')

# barra de cores
Map.add_colorbar(param_vis_temp, label='Temperatura (°C)', orientation='horizontal')
Map.add_colorbar(param_vis_frp, label='FRP (MW)', orientation='horizontal')

# adiciona legenda manual para o Fire Mask
Map.add_legend(legend_dict={'Fogo Ativo': 'skyblue'}, title='Fire Mask')

# exibe na tela
Map
```

#**PARTE 4)**: Visualização das Cicatrizes das Queimadas Através da Composição de Bandas do [Sentinel-2](https://developers.google.com/earth-engine/datasets/catalog/COPERNICUS_S2_SR_HARMONIZED?hl=pt-br)

![Texto alternativo](https://raw.githubusercontent.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_NOTTUS_2026/main/04_logos/image-20230517-132224.png)

O [Sentinel-2](https://space.oscar.wmo.int/satelliteprogrammes/view/sentinel_2) é uma constelação de satélites de observação da Terra desenvolvida pela Agência Espacial Europeia (ESA) como parte do programa Copernicus. Os satélites que compoêm o Sentinel-2 possuem o sensor Multi-Spectral Imager (MSI), que é um sensor de imagem multiespectral de alta resolução e ampla cobertura que dá suporte aos estudos de Monitoramento Terrestre do programa Copernicus. Sua combinação de bandas espectrais permite uma ampla gama de usos, incluindo o monitoramento da cobertura vegetal, do solo e da água, além da observação de vias navegáveis ​​interiores e áreas costeiras. O sensor MSI do Sentinel-2 possui 13 bandas espectrais abrangendo desde 443,9 a 2.202 nm, com diferentes resoluções espaciais: quatro bandas com resolução de 10 metros, seis bandas com resolução de 20 metros e três bandas com resolução de 60 metros.

Trabalhar com os dados do Sentinel-2 tem inúmeras vantagens como: alta resolução espacial (10-60m), ampla cobertura espectral (13 bandas), alta frequência de revisita (5 dias com dois satélites Sentinel-2A e 2B), dados gratuitos e abertos e radiometria calibrada (dados prontos para análise quantitativa e consistência entre diferentes imagens).

Iremos analisar as cicatrizes das queimadas através dos dados de reflectância de superfície [Harmonized Sentinel-2 MSI: MultiSpectral Instrument, Level-2A (SR)](https://developers.google.com/earth-engine/datasets/catalog/COPERNICUS_S2_SR_HARMONIZED?hl=pt-br) disponibilizados pelo GEE. O período dos dados do Sentinel-2 é desde 28 de março de 2017 até o momento presente, com frequencia temporal de 5 dias.

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/sentinel2.png?raw=true)

Para avaliar visualmente as características das queimadas podemos fazer o uso de composição de bandas. A composição de bandas permite combinar diferentes bandas espectrais para criar imagens coloridas com propósitos específicos.

A composição mais comum utilizada em Sensoriamento Remoto para ver imagens é a conhecida como RGB (Red-Green-Blue). Ela é uma representação mais próxima do que o olho humano veria. Para o Sentinel-2 elas correspondem as seguintes bandas:
1. `R`: Banda 4 (Vermelho - 665 nm)
2. `G`: Banda 3 (Verde - 560 nm)
3. `B`: Banda 2 (Azul - 490 nm)

Na composição RGB as cicatrizes de queimadas aparecem como manchas escuras/cinzas a pretas. Além disso:
- Solo exposto após uma queimada apresenta-se mais escuro que a vegetação.
- Em casos de fogo ativo, pode-se observar fumaça (branca a cinza).
- As cinzas/fuligens podem cobrir a vegetação remanescente.

Porém, é importante salientar que algumas vezes pode ser difícil distinguir entre solo natural e área queimada.



Baseado na figura abaixo, definiremos dois períodos para comparar os efeitos das queimadas numa determinada região do Estado de São Paulo:

1. Período `pré-fogo`: 19/agosto/2024
2. Período `pós-fogo`: 24/agosto/2024

Os dados do Sentinel-2 possuem um fator de escala, então multiplicaremos os dados por 0.0001 (veja no [site](https://developers.google.com/earth-engine/datasets/catalog/COPERNICUS_S2_SR_HARMONIZED?hl=pt-br) dos dados). Além disso selecionaremos apenas imagens com menos de 20% de nuvens na região de interesse. Em seguida plotaremos as imagens pré e pós-fogo no mapa interativo.

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/FOCOS_PRE_POS.png?raw=true)

```python
#=================================================================================================#
#                                  CARREGA OS DADOS
#=================================================================================================#
# datas
data_pre_1, data_pre_2 = '2024-08-19', '2024-08-20' # Imagem antes da queimada (pré-fogo)
data_pos_1, data_pos_2 = '2024-08-24', '2024-08-25' # Imagem depois da queimada (pós-fogo)

# região de interesse
regiao_recorte = ee.Geometry.Rectangle([-49.8169, -20.2338, -49.7098, -20.1358]) # [lonmin, latmin, lonmax, latmax]

# carrega os dados
sentinel2 = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED') \
              .filterBounds(regiao_recorte) \
              .filterDate('2024-08-19','2024-08-25') \
              .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 20)) \
              .map(lambda img: img.multiply(0.0001).copyProperties(img, img.propertyNames()))

# seleciona a imagem pré-fogo
imagem_pre = sentinel2.filterDate(data_pre_1, data_pre_2).mean().clip(regiao_recorte)

# seleciona a imagem pós-fogo
imagem_pos = sentinel2.filterDate(data_pos_1, data_pos_2).mean().clip(regiao_recorte)

#=================================================================================================#
#                                    PLOTA FIGURA
#=================================================================================================#
# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(regiao_recorte, 13)

# parâmetros de visualização
param_vis_rgb = {'bands': ['B4','B3','B2'], 'min': 0.03, 'max': 0.19}

# plota imagens
Map.addLayer(imagem_pre, param_vis_rgb, 'Pré-Fogo')
Map.addLayer(imagem_pos, param_vis_rgb, 'Pós-Fogo')

# exibe na tela
Map
```

Para realizar uma comparação mais detalhada entre a imagem pré e pós-fogo podemos utilizar a função [**split map**](https://geemap.org/geemap/#geemap.geemap.Map.split_map) do geemap. Esta função mostra num único painel as duas imagens, assim podemos arrastar para esquerda e direita e observar a mudança visual que ocorreu na  área que foi queimada. Veja um exemplo [aqui](https://geemap.org/notebooks/04_split_panel_map/).

```python
#========================================================================================================================#
```
```python
#                                                    INSTALA BIBLIOTECAS
#========================================================================================================================#
!pip install -q rasterio cartopy

#========================================================================================================================#
#                                             CRIA DIRETÓRIO DE SAÍDA
#========================================================================================================================#
dir_output = '/content/output/Parte_4'
os.makedirs(dir_output, exist_ok=True)

#========================================================================================================================#
#                                               IMPORTA BIBLIOTECAS
#========================================================================================================================#
import rasterio
import numpy as np
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
from cartopy.mpl.gridliner import LONGITUDE_FORMATTER, LATITUDE_FORMATTER

#========================================================================================================================#
#                                        FUNÇÃO DE LEITURA DA IMAGEM TIF
#========================================================================================================================#
# função que faz a leitura da imagem TIF
def read_tif(file):

    # leitura da imagem TIF
    with rasterio.open(file) as src:

        # leitura das bandas
        b1 = src.read(1)
        b2 = src.read(2)
        b3 = src.read(3)

        # combinar as bandas em uma imagem. Normalizar os valores se necessário (dependendo do tipo de dados)
        if b1.dtype != np.uint8:

            # normalizar para 0-255 se for float ou outro tipo
            b1_norm = ((b1 - b1.min()) / (b1.max() - b1.min()) * 255).astype(np.uint8)
            b2_norm = ((b2 - b2.min()) / (b2.max() - b2.min()) * 255).astype(np.uint8)
            b3_norm = ((b3 - b3.min()) / (b3.max() - b3.min()) * 255).astype(np.uint8)
            combinado_image = np.dstack((b1_norm, b2_norm, b3_norm))

        else:

            combinado_image = np.dstack((b1, b2, b3))

        # obter as informações de georreferenciamento
        transform = src.transform
        bounds = src.bounds
        crs = src.crs

    return combinado_image, transform, bounds, crs

#========================================================================================================================#
#                                               PLOTA FIGURAS
#========================================================================================================================#
# limites da imagem
lonmin, lonmax, latmin, latmax = -49.8169, -49.7098, -20.2338, -20.1358

# cria a figura
fig, axes = plt.subplots(2, 2, figsize=(12, 10), subplot_kw={'projection': ccrs.PlateCarree()})

#--------------------------------------------------------------#
#                   LOOP DAS IMAGENS
#--------------------------------------------------------------#
# nomes dos arquivos
files = ['Parte_4_Fig_1a_rgb_pre_fogo.tif',
         'Parte_4_Fig_1b_rgb_pos_fogo.tif',
         'Parte_4_Fig_2a_falsacor_pre_fogo.tif',
         'Parte_4_Fig_2b_falsacor_pos_fogo.tif']

# títulos das figuras
titulos = ['RGB: Pré-fogo', 'RGB: Pós-fogo', 'Falsa Cor: Pré-fogo', 'Falsa Cor: Pós-fogo']

# loop nas fguras
for i, ax in enumerate(axes.flat):

    # leitura da imagem TIF
    path = f'{dir_output}/{files[i]}'
    imagem, transform, bounds, crs = read_tif(path)

    # extrai os limites da imagem
    extent = [bounds.left, bounds.right, bounds.bottom, bounds.top]

    # plota imagem
    ax.imshow(imagem, origin='upper', extent=extent, transform=ccrs.PlateCarree())

    # define as extensão da imagem
    ax.set_extent([lonmin, lonmax, latmin, latmax], crs=ccrs.PlateCarree())

    # adiciona grade
    gl = ax.gridlines(draw_labels=True, xlocs=np.arange(-180, 180, 0.05), ylocs=np.arange(-90, 90, 0.05), linewidth=0.5, color='gray', alpha=0.5)
    gl.top_labels = False
    gl.right_labels = False
    gl.xformatter = LONGITUDE_FORMATTER
    gl.yformatter = LATITUDE_FORMATTER

    # plota subtítulo
    ax.set_title(titulos[i], fontsize=10)

# plota subtítulo geral da figura
plt.suptitle('Sentinel-2', fontsize=14)
plt.tight_layout()

# salva figura
plt.savefig(f'{dir_output}/Parte_4_Fig_3_pre_e_pos_fogo.jpg', dpi=300, bbox_inches="tight")

# mostra figura na tela
plt.show()
```

#**PARTE 5)**: Normalized Difference Vegetation Index (NDVI)

![Texto alternativo](https://raw.githubusercontent.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_NOTTUS_2026/main/04_logos/Classificação_Valores_NDVI_PTBR.png)

Como as queimadas causam mudanças nas características do solo, podemos utilizar o `Normalized Difference Vegetation Index (NDVI)` para verificar esta mudança. O NDVI emprega a banda do infravermelho próximo (0,8 µm, NIR) e vermelho (0,6 µm, R) para realçar as características da vegetação numa determinada região. A seguinte equação é utilizada para o cálculo do NDVI:

---

$$
NDVI = \frac{NIR − R}{NIR + R}
$$

---

Em que:
 - `NIR:` refletância no canal de 0,8 µm. Banda B8-NIR (835.1nm (S2A) / 833nm (S2B)) do Sentinel-2 com 10 metros.

 - `R:` refletância no canal de 0,6 µm. Banda B4-Red (664.5nm (S2A) / 665nm (S2B)) do Sentinel-2 com 10 metros.

---

*Os valores de NDVI tipicamente concentram-se entre -1 a 1, sendo que valores baixos indicam solo nu/pouca vegetação e valores mais altos indicam vegetação sadia. Como exemplo veja a figura acima.*

Importante ressaltar que o objetivo do NDVI não é identificar área com queimadas, mas sim as características da vegetação. Então aqui queremos apenas observar algum indicativo na mudança da refletância da superfície ocorrida após o fogo na vegetação.

```python
#=================================================================================================#
#                                    DEFINE A FUNÇÃO QUE CALCULA O NDVI
#=================================================================================================#
def ndvi(colecao):

    """
    Calcula o NDVI para uma coleção de imagens do Sentinel-2,
    mantendo todas as bandas originais.

    Parâmetros:
    -----------
    colecao : ee.ImageCollection
        Coleção de imagens do Sentinel-2

    Retorna:
    --------
    ee.ImageCollection
        Coleção com todas as bandas originais + banda NDVI adicionada
    """

    # Função para calcular e adicionar NDVI a cada imagem
    def adicionar_ndvi(imagem):

        """
        Calcula NDVI e adiciona como nova banda à imagem.

        NDVI = (NIR - Red) / (NIR + Red)
        Sentinel-2: B8 = NIR, B4 = Red
        """
        # Calcula o índice NDVI
        ndvi = imagem.normalizedDifference(['B8', 'B4']).rename('NDVI')

        # Adiciona a banda NDVI mantendo todas as bandas originais
        return imagem.addBands(ndvi)

    # Aplica a função a todas as imagens da coleção
    return colecao.map(adicionar_ndvi)

#=================================================================================================#
#                                    CARREGA OS DADOS
#=================================================================================================#
# datas
data_pre_1, data_pre_2 = '2024-08-19', '2024-08-20' # Imagem antes da queimada (pré-fogo)
data_pos_1, data_pos_2 = '2024-08-24', '2024-08-25' # Imagem depois da queimada (pós-fogo)

# região de interesse
regiao_recorte = ee.Geometry.Rectangle([-49.8169, -20.2338, -49.7098, -20.1358])


# carrega os dados
sentinel2 = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED') \
              .filterBounds(regiao_recorte) \
              .filterDate('2024-08-01','2024-09-01') \
              .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 20)) \
              .map(lambda img: img.multiply(0.0001).copyProperties(img, img.propertyNames()))

#=================================================================================================#
#                                    CALCULA O NDVI
#=================================================================================================#
sentinel2_com_ndvi = ndvi(sentinel2)

#=================================================================================================#
#                                SELECIONA IMAGEM PRÉ E PÓS-FOGO
#=================================================================================================#
# seleciona as imagens com NDVI pré-fogo
imagem_pre = sentinel2_com_ndvi.filterDate(data_pre_1, data_pre_2).mean().clip(regiao_recorte)

# seleciona as imagens com NDVI pós-fogo
imagem_pos = sentinel2_com_ndvi.filterDate(data_pos_1, data_pos_2).mean().clip(regiao_recorte)

# as imagens possuem:
# 1. Todas as bandas originais do Sentinel-2 (B1, B2, B3, B4, B5, B6, B7, B8, B8A, B11, B12)
# 2. A banda NDVI calculada

# NDVI pré-fogo
ndvi_pre = imagem_pre.select('NDVI')

# NDVI pró-fogo
ndvi_pos = imagem_pos.select('NDVI')

#=================================================================================================#
#                                    PLOTA FIGURA
#=================================================================================================#
# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(regiao_recorte, 13)

# parâmetros de visualização
param_vis_ndvi = {'min':-0.5, 'max':1, 'palette': ['#8B0000',  # marrom avermelhado (áreas queimadas)
                                                   '#D2691E',  # marrom (solo exposto pós-fogo)
                                                   '#FFD700',  # amarelo (vegetação em recuperação)
                                                   '#9ACD32',  # verde amarelado
                                                   '#32CD32',  # verde (vegetação saudável)
                                                   '#006400'   # verde escuro (vegetação muito saudável)
                                                     ]}

# define o mapa da esquerda (pré-fogo)
left_layer = geemap.ee_tile_layer(ndvi_pre, param_vis_ndvi, 'pré-fogo')

# define o mapa da direita (pós-fogo)
right_layer = geemap.ee_tile_layer(ndvi_pos, param_vis_ndvi, 'pós-fogo')

# mostra mapa
Map.split_map(left_layer, right_layer, left_label='NDVI: Pré-Fogo', right_label='NDVI: Pós-Fogo')

# exibe na tela
Map
```

#**PARTE 6)**: Normalized Burn Ratio (NBR)

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/curva_area_queimada.png?raw=true)

Nas etapas anteriores da nossa aula nós utilizamos as composições RGB, Falsa cor e o índice espectral NDVI para identificar visualmente uma região que teve queimadas. Porém, torna-se importante neste momento aplicar um índice espectral específico para queimadas e calcular a área (em km²) que foi efetivamente queimada pelos incêndios.

Para isto utilizaremos o índice [Normalized Burn Ratio (NBR)](https://un-spider.org/advisory-support/recommended-practices/recommended-practice-burn-severity/in-detail/normalized-burn-ratio). A seguir temos uma explicação do NBR e como podemos calculá-lo.

---
- `Normalized Burn Ratio (NBR)`:
---

> O cálculo do `Normalized Burn Ratio (NBR)` identifica as áreas que foram queimadas e o grau de severidade da queimada usando imagens de satélite, com a fórmula:

$$
NBR = \frac{NIR - SWIR}{NIR + SWIR}
$$

Em que:
 - `NIR:` refletância no infravermelho próximo, canal de 0,83 µm  
 - `SWIR:` refletância no infravermelho curto, canal de 2,1 µm

> Para o [Sentinel-2](https://developers.google.com/earth-engine/datasets/catalog/COPERNICUS_S2_SR_HARMONIZED?hl=pt-br) serão os canais `NIR:` B8 (0,835 e 0,833µm) e `SWIR:` B12 (2,2 e 2,1 µm).

> Interpretação dos valores:
- Os valores variam de -1 a 1:
    1. Valores positivos geralmente indicam vegetação saudável.
    2. Valores próximos ou abaixo de zero indicam solo queimado.

---
- `Difference Normalized Burn Ratio (dNBR)`:
---


> Agora para avaliar a severidade da queimada podemos utilizar o `Difference Normalized Burn Ratio (dNBR)`. O dNBR é calculado através da diferença entre a imagem do NBR pré-fogo e pós-fogo:

$$
dNBR = NBR(pré) - NBR(pós)
$$

> Interpretação dos valores:
1.  Valores positivos mais elevados indicam maior gravidade da queimada.

Mais recentemente [Alcaraz et al. (2022)](https://www.mdpi.com/2072-4292/14/7/1727) desenvolveram o Normalized Burn Ratio Plus (NBR+). O NBR+ é um aprimoramento do NBR clássico. Ao incorporar as bandas Azul (B2) e Verde (B3) do Sentinel-2, juntamente com as bandas NIR (B8A) e SWIR2 (B12), o NBR+ suprime inerentemente falsos positivos de corpos d'água e nuvens sem a necessidade de camadas de mascaramento externas. Utiliza-se a seguinte equação:

$$
NBR+ = \frac{B12 - B8A - B3 - B2}{B12 + B8A + B3 + B2}
$$

Em que:
 - `B2:` refletância no azul, canal de 496,6 nm (S2A) / 492,1 nm (S2B)
 - `B3:` refletância no verde, canal de 560 nm (S2A) / 559 nm (S2B)
 - `B8A:` refletância no infravermelho próximo (NIR), canal de 864,8 nm (S2A) / 864 nm (S2B)
 - `B12:` refletância no SWIR 2, canal de 2202.4nm (S2A) / 2185.7nm (S2B)


1. Corpos d'água possuem forte refletância em comprimentos de onda azul-esverdeados, assim a inclusão das bandas B2 e B3 na subtração leva os valores de NBR+ de pixels de água para valores negativos.

2. Nuvens também produzem valores de NBR+ fortemente negativos devido à sua baixa refletância na banda SWIR2 (B12) em relação às outras três bandas.

3. Áreas queimadas, caracterizadas por alta refletância no infravermelho curto (SWIR), produzem valores de NBR+ distintamente altos e positivos. Essa separação espectral permite a identificação de áreas queimadas sem depender de máscaras externas de água ou nuvens.

#**PARTE 7)**: Analytical Burned Area Index (ABAI)

O Analytical Burned Area Index (ABAI) foi desenvolvido por [Wu et al. (2022)](https://www.mdpi.com/1999-4907/13/11/1787) e possui a vantagem em discriminar melhor entre solo exposto nativo, areia, nuvens e até água em alguns casos das áreas queimadas. O ABAI foi construído através de uma análise estatística das assinaturas espectrais de diversas classes (como vegetação, água, solo nu, sombra, construção e nuvens) e aplicando-se uma metodologia baseada em otimização multiobjetivo. Assim foi desenvolvido funções matemáticas específicas para cada classe confundível, buscando maximizar a separação espectral entre áreas queimadas e os demais alvos. Os índices mais tradicionais utilizam apenas duas bandas, enquanto o ABAI é uma combinação de três bandas: infravermelho de ondas curtas (SWIR1 e SWIR2) e a banda do verde (B3). Utiliza-se a seguinte equação:

$$
ABAI = \frac{3 * SWIR2 - 2 * SWIR1 - 3 * GREEN}{3 * SWIR2 + 2 * SWIR1 + 3 * GREEN}
$$

Em que:
 - `GREEN (B3):` refletância no verde, canal de 560 nm (S2A) / 559 nm (S2B) - 10 metros.
 - `SWIR1 (B11):` refletância no infravermelho curto, canal de 1,613,7 nm (S2A) / 1,610,4 nm (S2B) - 20 metros.
 - `SWIR2 (B12):` refletância no infravermelho curto, canal de 2.202,4 nm (S2A) / 2.185,7 nm (S2B) - 20 metros.

Assim, valores positivos estão associados as áreas queimadas e negativos a alvos como água e sombra, eliminando a necessidade de máscaras complementares.

Os resultados experimentais demonstraram que a inclusão da banda do verde no índice espectral resultou em boa aplicabilidade na detecção de áreas queimadas, onde o ABAI consegue evitar a maior parte da confusão causada por sombras ou águas rasas. Portanto, com uma acurácia superior a 90%, o ABAI se mostra altamente promissor para aplicações em gestão ambiental, planejamento territorial e resposta a desastres.

De modo semelhante ao realizado com o índice NBR aplicaremos para o ABAI. Lembrando que usaremos uma imagem pré-fogo e outra pós-fogo do satélite Sentinel-2.

```python
#=================================================================================================#
#                                    CARREGA OS DADOS
#=================================================================================================#
# datas
data_pre_1, data_pre_2 = '2024-08-19', '2024-08-20' # data antes da queimada (pré-fogo)
data_pos_1, data_pos_2 = '2024-08-24', '2024-08-25' # data depois da queimada (pós-fogo)

# região de interesse
regiao_recorte = ee.Geometry.Rectangle([-49.8169, -20.2338, -49.7098, -20.1358])

# carrega os dados
sentinel2 = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED') \
              .filterBounds(regiao_recorte) \
              .filterDate('2024-08-01','2024-09-01') \
              .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 20)) \
              .map(lambda img: img.multiply(0.0001).copyProperties(img, img.propertyNames()))


# seleciona a imagem pré-fogo
imagem_pre = sentinel2.filterDate(data_pre_1, data_pre_2).mean().clip(regiao_recorte)

# seleciona a imagem pós-fogo
imagem_pos = sentinel2.filterDate(data_pos_1, data_pos_2).mean().clip(regiao_recorte)

#=================================================================================================#
#                                  CÁLCULO DO ABAI e dABAI
#=================================================================================================#
# função para calcular NBRSWIR (Sentinel-2: SWIR1=B11, SWIR2=B12)
def abai(image):
    return image.expression('(3 * SWIR2 - 2 * SWIR1 - 3 * GREEN) / (3 * SWIR2 + 2 * SWIR1 + 3 * GREEN)',
                            {'SWIR2': image.select('B12'),
                             'SWIR1': image.select('B11'),
                             'GREEN': image.select('B3')}).rename('ABAI')

# ABAI pré-fogo
abai_pre = abai(imagem_pre)

# ABAI pós-fogo
abai_pos = abai(imagem_pos)

# calcula o dABAI
dabai = abai_pre.subtract(abai_pos).rename('dABAI')

#=================================================================================================#
#                                 APLICA A MASCARA NA ÁREA QUEIMADA
#=================================================================================================#
# limiares
limiar_min, limiar_max = -1.0, -0.12

# seleciona regiões onde a condição é válida
mascara_area_queimada = dabai.gte(limiar_min).And(dabai.lte(limiar_max))

# aplica a mascara aos dados
area_queimada_mascarada = dabai.updateMask(mascara_area_queimada)

# coloca 1 nos pixels que obdecem o critério e 0 onde não obdece o critério
abai_binario = area_queimada_mascarada.multiply(0).add(1)

#=================================================================================================#
#                                   CÁLCULO DA ÁREA (KM²)
#=================================================================================================#
# cálcula a área de cada pixel em km²:
# 1. Multiplica pela área do pixel em m²
# 2. Divide por 1.000.000 para converter para km²
# Note: mascara_area_queimada was commented out, so it was not defined. Using binaryABAI for calculation.
calculo_area_queimada_km2 = abai_binario.multiply(ee.Image.pixelArea()).divide(1000000)

# acumula todos os pixels em km², calculando assim a área total da queimada
soma_area_queimada_km2 = calculo_area_queimada_km2.reduceRegion(reducer=ee.Reducer.sum(),
                                                                geometry=regiao_recorte,
                                                                scale=20, # resolução das bandas SWIR do Sentinel-2
                                                                maxPixels=1e13,
                                                                bestEffort=True)

# obtém valor em km²
valor_area_queimada = soma_area_queimada_km2.get('dABAI').getInfo()
print(f"   📊 Área queimada: {valor_area_queimada:.1f} km²")


#=================================================================================================#
#                                    PLOTA FIGURA
#=================================================================================================#
# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(regiao_recorte, 13)

# parâmetros de visualização
param_vis_abai = {'min': -1.0, 'max': 1.0, 'palette': ['blue', 'green', 'orange', 'yellow', 'red']}
param_vis_dabai = {'min': -1.0, 'max': 1.0, 'palette': ['blue', 'green', 'orange', 'yellow', 'red']}
param_vis_queimada = {'palette': ['brown']}

# plota imagens
Map.addLayer(abai_pre, param_vis_abai, 'ABAI Pré-Fogo')
Map.addLayer(abai_pos, param_vis_abai, 'ABAI Pós-Fogo')
Map.addLayer(abai_pos, param_vis_dabai, 'dABAI')
Map.addLayer(abai_binario, param_vis_queimada, 'Área Queimada')

# exibe na tela
Map
```

Iremos utilizar como referência a latitude de -21.1775° e longitude de -47.8111° do município e aplicar com um raio de abrangência (buffer) de 1000 m para a série histórica de 1984 à 2026.

```python
#=============================================================================================#
#                                  IMPORTA BIBLIOTECAS
#=============================================================================================#
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np

#=============================================================================================#
#                               CRIA DIRETÓRIO DE SAÍDA
#=============================================================================================#
dir_output = '/content/output/Parte_8'
os.makedirs(dir_output, exist_ok=True)

#=============================================================================================#
#                                  CARREGA OS DADOS
#=============================================================================================#
# define a região: buffer de 1000 metros
regiao = ee.Geometry.Point([-47.8111, -21.1775]).buffer(1000) # (longitude, latitude) com buffer de 1000 metros

# carrega os dados
nbr = ee.ImageCollection('LANDSAT/COMPOSITES/C02/T1_L2_32DAY_NBR') \
        .filterDate('1984', '2027') \
        .filterBounds(regiao) \
        .select('NBR')

#=============================================================================================#
#                                 EXTRAI SÉRIE TEMPORAL
#=============================================================================================#
# função para extrair a série temporal média
def extrair_serie_temporal(colecao, regiao, escala=30): # 30 metros

    def processar_imagem(imagem):

        data = ee.Date(imagem.get('system:time_start')).format('YYYY-MM-dd')

        nbr_medio = imagem.reduceRegion(reducer = ee.Reducer.mean(),
                                        geometry = regiao,
                                        scale = escala,
                                        maxPixels = 1e9).get('NBR')

        return ee.Feature(None, {'Data': data, 'NBR': nbr_medio})

    return colecao.map(processar_imagem)

# extrai a série temporal
serie_temporal = extrair_serie_temporal(nbr, regiao)

# elimina os valores de datas/NBR com "NaN"
serie_temporal = serie_temporal.filter(ee.Filter.notNull(['NBR']))

# extrai as "DATAS" da FeatureCollection
datas = serie_temporal.aggregate_array('Data').getInfo()

# extrai os "DADOS de NBR" da FeatureCollection
valores = serie_temporal.aggregate_array('NBR').getInfo()

# converte para DataFrame
df = pd.DataFrame({'Data': pd.to_datetime(datas), 'NBR': valores}).sort_values('Data').reset_index(drop=True)

# remove valores nulos (caso existam)
df = df.dropna()

#=============================================================================================#
#                                  PLOTA FIGURA
#=============================================================================================#
# cria a moldura da figura
plt.figure(figsize=(14, 7))

# separa os dados por período: dados históricos (todos exceto 2024) - incluindo 2025 e 2026
df_historico = df[df['Data'].dt.year != 2024]

# separa os dados por período: dados de 2024
df_2024 = df[df['Data'].dt.year == 2024]

# plota curvas com cores diferentes: curva histórica (azul) - inclui 1984-2023, 2025 e 2026
plt.plot(df_historico['Data'], df_historico['NBR'], 'b-', linewidth=1.5, alpha=0.6, label='Histórico')

# plota curvas com cores diferentes: curva de 2024 (vermelho)
plt.plot(df_2024['Data'], df_2024['NBR'], 'r-', linewidth=2.5, alpha=0.9, label='2024')

# cria uma paleta de cores baseada no ano para os pontos
anos = df['Data'].dt.year
cores = []
for ano in anos:
    if ano == 2024:
        cores.append('red')
    else:
        cores.append('blue')  # Todos os outros anos em azul (incluindo 2025 e 2026)

# plota os pontos com cores correspondentes
plt.scatter(df['Data'], df['NBR'], c=cores, s=25, alpha=0.7, edgecolors='black', linewidth=0.5)

# destaque para os pontos de 2024 (maiores e com borda)
plt.scatter(df_2024['Data'], df_2024['NBR'], c='red', s=80, alpha=0.9,
            edgecolors='yellow', linewidth=1.5, label='Pontos 2024', zorder=5)

# adiciona anotações para identificar 2025 e 2026
# encontra os dados de 2025 e 2026
df_2025 = df[df['Data'].dt.year == 2025]
df_2026 = df[df['Data'].dt.year == 2026]

# adiciona anotações para indicar 2025 e 2026
if not df_2025.empty:

    # pega o ponto médio de 2025 para anotação
    idx_2025 = len(df_2025) // 2
    if idx_2025 < len(df_2025):
        data_2025 = df_2025.iloc[idx_2025]['Data']
        nbr_2025 = df_2025.iloc[idx_2025]['NBR']
        plt.annotate('2025', xy=(data_2025, nbr_2025),
                     xytext=(10, 10), textcoords='offset points',
                     fontsize=9, color='blue', alpha=0.7,
                     bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.7))

if not df_2026.empty:

    # pega o ponto médio de 2026 para anotação
    idx_2026 = len(df_2026) // 2
    if idx_2026 < len(df_2026):
        data_2026 = df_2026.iloc[idx_2026]['Data']
        nbr_2026 = df_2026.iloc[idx_2026]['NBR']
        plt.annotate('2026', xy=(data_2026, nbr_2026),
                     xytext=(10, 10), textcoords='offset points',
                     fontsize=9, color='blue', alpha=0.7,
                     bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.7))

# configuração dos eixos do gráfico
plt.title('Série Temporal de NBR - LANDSAT: 1984-2026\nDestaque para o Ano de 2024', fontsize=14, fontweight='bold')
plt.ylabel('NBR', fontsize=12)
plt.xlabel('Data', fontsize=12)

# adiciona grades
plt.grid(True, linestyle='--', alpha=0.3)

# adiciona legenda
plt.legend(loc='best', fontsize=10, framealpha=0.9)

# ajusta os limites do eixo y para melhor visualização
if not df.empty:
    y_min = df['NBR'].min() - 0.05
    y_max = df['NBR'].max() + 0.05
    plt.ylim(y_min, y_max)

# adiciona uma linha vertical para marcar o início e fim de 2024
plt.axvline(x=pd.Timestamp('2024-01-01'), color='gray', linestyle='--', alpha=1, linewidth=0.8)
plt.axvline(x=pd.Timestamp('2025-01-01'), color='gray', linestyle='--', alpha=1, linewidth=0.8)

# recorta excesso de borda
plt.tight_layout()

# salva a figura
plt.savefig(f'{dir_output}/Parte_8_serie_temporal_1984-2026_NBR_LANDSAT.png', dpi=300, bbox_inches='tight')

# mostra a figura
plt.show()

#=============================================================================================#
#                          ESTATÍSTICAS DO ANO DE 2024
#=============================================================================================#
print("\n" + "="*60)
print("ESTATÍSTICAS DO ANO DE 2024")
print("="*60)

if not df_2024.empty:
    print(f"Total de observações em 2024: {len(df_2024)}")
    print(f"NBR médio em 2024: {df_2024['NBR'].mean():.4f}")
    print(f"NBR máximo em 2024: {df_2024['NBR'].max():.4f}")
    print(f"NBR mínimo em 2024: {df_2024['NBR'].min():.4f}")
    print(f"Desvio padrão em 2024: {df_2024['NBR'].std():.4f}")
    print(f"Período: {df_2024['Data'].min().strftime('%Y-%m-%d')} a {df_2024['Data'].max().strftime('%Y-%m-%d')}")

    # Comparação com média histórica (excluindo 2024)
    if not df_historico.empty:
        media_historica = df_historico['NBR'].mean()
        print(f"\nMédia histórica (excluindo 2024): {media_historica:.4f}")
        print(f"Diferença 2024 vs Histórico: {df_2024['NBR'].mean() - media_historica:.4f}")

        # Tendência
        if df_2024['NBR'].mean() > media_historica:
            print(f"→ 2024 apresenta NBR SUPERIOR à média histórica")
        else:
            print(f"→ 2024 apresenta NBR INFERIOR à média histórica")
else:
    print("Nenhum dado disponível para o ano de 2024")
```

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/modis.png?raw=true)

Existem diversos produtos relacionados a queimadas gerados a partir dos dados obtidos pelo sensor [MODIS](https://space.oscar.wmo.int/instruments/view/modis) que esta abordo dos satélites [AQUA](https://space.oscar.wmo.int/satellites/view/aqua) e [TERRA](https://space.oscar.wmo.int/satellites/view/terra). Basicamente, citaremos três produtos que são (MOD14A1, MYD14A1 e MCD64A1):

1. `MOD_14A1 (active fire product)`: [Terra Thermal Anomalies & Fire Daily Global 1km](https://developers.google.com/earth-engine/datasets/catalog/MODIS_061_MOD14A1?hl=pt-br). Dados globais de *Potência radiativa máxima do incêndio* (MaxFRP) desde 2000 com resolução temporal e espacial de 1 dia e 1km, respectivamente obtido pelo satélite TERRA.

2. `MYD_14A1 (active fire product)`: [Aqua Thermal Anomalies & Fire Daily Global 1km](https://developers.google.com/earth-engine/datasets/catalog/MODIS_061_MYD14A1?hl=pt-br). Dados globais de *Potência radiativa máxima do incêndio* (MaxFRP) desde 2002 com resolução temporal e espacial de 1 dia e 1km, respectivamente obtido pelo satélite AQUA.

2. `MCD_64A1 (burned area product)`: [MODIS Burned Area Monthly Global 500m](https://developers.google.com/earth-engine/datasets/catalog/MODIS_061_MCD64A1?hl=pt-br). São os dados do dia do ano (conhecido como dia Juliano) que teve queimada. Valores possíveis: 0 (não queimado), 1 a 366 (dia juliano aproximado da queimada). Os dados são desde  desde 2000.

No exemplo a seguir plotaremos o mapa interativo do dia juliano que teve queimada fornecido pelo produto `MCD_64A1 (burned area product)`. Usaremos como referência o período de 2024-08-01 à 2024-10-01.

```python
#=================================================================================================#
#                                    CARREGA OS DADOS
#=================================================================================================#
# carrega os dados "MYD_14A1 (active fire product)", possuem os dados de FireMask e MaxFRP
# são os dados Aqua Thermal Anomalies & Fire Daily Global 1km

# região de interesse
estado_sp = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo'))

# Aqua Thermal Anomalies & Fire Daily Global 1km: FRP
MYD14A1_active_fire_FRP = ee.ImageCollection('MODIS/061/MYD14A1') \
                            .filterBounds(estado_sp) \
                            .filterDate('2024-08-01','2024-10-01')  \
                            .select('MaxFRP') \
                            .map(lambda image: image.multiply(0.1).copyProperties(image, ['system:time_start']))

#=================================================================================================#
#                                 PLOTA MAPA INTERATIVO
#=================================================================================================#
# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(estado_sp, 7)

# parâmetros de visualização
vis_FRP = {'min': 0.0, 'max': 1500, 'palette': ['black', 'blue', 'pink', 'red']}

# mapas
Map.addLayer(MYD14A1_active_fire_FRP.mean().clip(estado_sp), vis_FRP, 'MaxFRP')

# contorno do estado de SP
style = {'color': 'yellow', 'fillColor': '00000000'}
Map.addLayer(estado_sp.style(**style), {}, 'SP')

# barra de cores
Map.add_colorbar(vis_FRP, label='MaxFRP (MW)', orientation='horizontal')

# exibe na tela
Map
```

#**PARTE 10)**: Produto [MCD19A2](https://developers.google.com/earth-engine/datasets/catalog/MODIS_061_MCD19A2_GRANULES?hl=pt-br#bands) de Aerossol dos Satélites AQUA e TERRA

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/MCD19A2_061.png?raw=true)

A profundidade óptica de aerossol (AOD) é uma medida da quantidade de aerossois dispersa na atmosfera. Como as queimadas acabam tendo como resultado a produção de aerossois, podemos utilizar as estimativas dos sensores dos satélites para avaliar a quantidade de poluição numa determinada região. Os satélites AQUA e TERRA possuem o produto [MCD19A2 V6.1](https://developers.google.com/earth-engine/datasets/catalog/MODIS_061_MCD19A2_GRANULES?hl=pt-br#bands) que fornece diariamente com resolução espacial de 1km a AOD na banda do azul (0,47 μm, Optical_Depth_047) e verde (0,55 μm, Optical_Depth_047) do sensor MODIS.

Como exemplo iremos plotar o mapa de AOD de 0,55 μm para o Estado de São Paulo para o período das queimadas que estamos estudando. Iremos utilizar o período entre os dias 22 e 24 de agosto de 2024 e extrair o máximo valor de AOD entre estes dias. Além disso precisaremos aplicar o fator de escala (x 0.001) nos dados antes de utilizá-los.

```python
#=================================================================================================#
#                                    CARREGA OS DADOS
#=================================================================================================#
# região de interesse
estado_sp = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo'))

# carrega os dados.
AOD55 = ee.ImageCollection('MODIS/061/MCD19A2_GRANULES') \
          .filterBounds(estado_sp) \
          .filterDate('2024-08-22','2024-08-25') \
          .select('Optical_Depth_055')\
          .map(lambda image: image.multiply(0.001).copyProperties(image, ['system:time_start']))

#=================================================================================================#
#                                 PLOTA MAPA INTERATIVO
#=================================================================================================#
# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(estado_sp, 7)

# parâmetros de visualização
vis = {'min': 0, 'max': 2, 'palette': ['black', 'blue', 'purple', 'cyan', 'green', 'yellow', 'red']}

# plota mapa
Map.addLayer(AOD55.max().clip(estado_sp), vis, 'AOD055')

# contorno da região
style1 = {'color': 'red', 'fillColor': '00000000'}
Map.addLayer(estado_sp.style(**style1), {}, 'Estado de SP')

# barra de cores
Map.add_colorbar_branca(colors=vis['palette'], vmin=vis['min'], vmax=vis['max'], layer_name='AOD')

# exibe na tela
Map
```

#**PARTE 11)**: Produtos de [Gases Amosféricos](https://developers.google.com/earth-engine/datasets/catalog/sentinel-5p?hl=pt-br) Estimados pelo Sentinel-5P

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/SENITNEL5P.png?raw=true)

O satélite [Sentinel-5 Precursor (5P)](https://space.oscar.wmo.int/satellites/view/sentinel_5p) fornece a cada 2 dias dados globais sobre a qualidade do ar. O sensor [Tropospheric Monitoring Instrument(TROPOMI)](https://space.oscar.wmo.int/instruments/view/tropomi) é um sensor multiespectral que registra a refletância de comprimentos de onda importantes para medir as concentrações atmosféricas de ozônio, metano, formaldeído, aerossol, monóxido de carbono, óxido de nitrogênio e dióxido de enxofre, bem como as características das nuvens. Os [dados](https://developers.google.com/earth-engine/datasets/catalog/sentinel-5p?hl=pt-br) possuem resolução espacial de 1.113,2 metros e estão disponiveis desde 2018 até o momento presente.

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/sentienl5p_co.png?raw=true)

Como exemplo iremos plotar o mapa de Monóxido de Carbono (CO) para o Estado de São Paulo para o período das queimadas que estamos estudando. Iremos utilizar o período entre os dias 22 e 24 de agosto de 2024 e extrair o máximo valor de CO entre estes dias. Além disso converteremos os dados de *mol/m${^2}$* para *Unidades Dobson (DU)*, multiplicando os dados originais por 2241.15.

```python
#=================================================================================================#
#                                    CARREGA OS DADOS
#=================================================================================================#
# região de interesse
estado_sp = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo'))

# carrega os dados.
S5P_co = ee.ImageCollection('COPERNICUS/S5P/NRTI/L3_CO') \
           .filterBounds(estado_sp) \
           .filterDate('2024-08-22','2024-08-25') \
           .select('CO_column_number_density')

#=================================================================================================#
#                                 PLOTA MAPA INTERATIVO
#=================================================================================================#
# cria a moldura do mapa
Map = geemap.Map()

# centraliza o mapa
Map.centerObject(estado_sp, 7)

# parâmetros de visualização
vis = {'min': 0, 'max': 150, 'palette': ['black', 'blue', 'purple', 'cyan', 'green', 'yellow', 'red']}

# plota mapa
Map.addLayer(S5P_co.max().clip(estado_sp).multiply(2241.15), vis, 'Mapa de CO (DU)')

# contorno da região
style1 = {'color': 'red', 'fillColor': '00000000'}
Map.addLayer(estado_sp.style(**style1), {}, 'Estado de SP')

# barra de cores
Map.add_colorbar_branca(colors=vis['palette'], vmin=vis['min'], vmax=vis['max'], layer_name='CO (DU)')

# exibe na tela
Map
```

```python
#========================================================================================================================#
#                                                    INSTALA BIBLIOTECAS
#========================================================================================================================#
!pip install -q xee cartopy ultraplot rasterio

#========================================================================================================================#
#                                                   IMPORTA BIBLIOTECAS
#========================================================================================================================#
import rasterio
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import ultraplot as uplt
import numpy as np
import xee
from xee import helpers
from shapely.geometry import shape
import xarray as xr
import os
import warnings
warnings.filterwarnings("ignore")

#========================================================================================================================#
#                                                 CRIA DIRETÓRIO DE SAÍDA
#========================================================================================================================#
dir_output = '/content/output/Parte_12'
os.makedirs(dir_output, exist_ok=True)

#========================================================================================================================#
#                                       CARREGA OS DADOS DO SENTINEL-2
#========================================================================================================================#
# datas
data_pre_1, data_pre_2 = '2024-08-19', '2024-08-20' # Imagem antes da queimada (pré-fogo)
data_pos_1, data_pos_2 = '2024-08-24', '2024-08-25' # Imagem depois da queimada (pós-fogo)

# região de interesse
regiao_recorte = ee.Geometry.Rectangle([-49.8169, -20.2338, -49.7098, -20.1358])

# carrega os dados
sentinel2 = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED') \
              .filterBounds(regiao_recorte) \
              .filterDate('2024-08-01','2024-09-01') \
              .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 20)) \
              .map(lambda img: img.multiply(0.0001).copyProperties(img, img.propertyNames()))

# seleciona a imagam pré-fogo
imagem_pre = sentinel2.filterDate(data_pre_1, data_pre_2).mean().clip(regiao_recorte)

# seleciona a imagam pós-fogo
imagem_pos = sentinel2.filterDate(data_pos_1, data_pos_2).mean().clip(regiao_recorte)

#========================================================================================================================#
#                                               1) FOCOS DE CALOR
#========================================================================================================================#
# intervalo de datas
datai, dataf = '2024-08-20', '2024-08-25'

#----------------------------------------------------------------#
#                         FIRMS
#----------------------------------------------------------------#
# FIRMS
firms = ee.ImageCollection('FIRMS') \
          .filterBounds(regiao_recorte) \
          .filterDate(datai,dataf) \
          .select('T21') \
          .map(lambda img: img.subtract(273.15).copyProperties(img, img.propertyNames()))

# converte "ee.Geometry" para "shapely geometry" explicitamente
shapely_geometry = shape(regiao_recorte.getInfo())

# latitude central da geometria
centroide = shapely_geometry.centroid
lat_media = centroide.y  # em graus

# calcula o grid_scale para 1000m
resolucao_metros = 1000.
metros_por_grau = 111320 * np.cos(np.radians(lat_media))
grid_scale_graus = resolucao_metros / metros_por_grau

# monta os parametros
grid_params = helpers.fit_geometry(geometry=shapely_geometry,
                                   grid_crs='EPSG:4326',
                                   grid_scale=(grid_scale_graus, -grid_scale_graus))

# converte para Dataset
ds_firms = xr.open_dataset(firms.mean(), engine = 'ee', **grid_params)

#----------------------------------------------------------------#
#                      VIIRS (NOAA-20)
#----------------------------------------------------------------#
# VIIRS (NOAA-20)
ti4_viirs_noaa20 = ee.ImageCollection('NASA/LANCE/NOAA20_VIIRS/C2') \
                     .filterBounds(regiao_recorte) \
                     .filterDate(datai,dataf) \
                     .select('Bright_ti4') \
                     .map(lambda img: img.subtract(273.15).copyProperties(img, img.propertyNames()))

# converte "ee.Geometry" para "shapely geometry" explicitamente
shapely_geometry = shape(regiao_recorte.getInfo())

# latitude central da geometria
centroide = shapely_geometry.centroid
lat_media = centroide.y  # em graus

# calcula o grid_scale para 375m
resolucao_metros = 375.
metros_por_grau = 111320 * np.cos(np.radians(lat_media))
grid_scale_graus = resolucao_metros / metros_por_grau

# monta os parametros
grid_params = helpers.fit_geometry(geometry=shapely_geometry,
                                   grid_crs='EPSG:4326',
                                   grid_scale=(grid_scale_graus, -grid_scale_graus))

# converte para Dataset
ds_ti4_viirs_noaa20 = xr.open_dataset(ti4_viirs_noaa20.mean(), engine = 'ee', **grid_params)

#----------------------------------------------------------------#
#                      PROFUNDIDADE ÓTICA
#----------------------------------------------------------------#
# Optical_Depth_055
AOD55 = ee.ImageCollection('MODIS/061/MCD19A2_GRANULES') \
          .filterBounds(regiao_recorte) \
          .filterDate(datai,dataf)  \
          .select('Optical_Depth_055')\
          .map(lambda image: image.multiply(0.001).copyProperties(image, ['system:time_start']))

# converte "ee.Geometry" para "shapely geometry" explicitamente
shapely_geometry = shape(regiao_recorte.getInfo())

# latitude central da geometria
centroide = shapely_geometry.centroid
lat_media = centroide.y  # em graus

# calcula o grid_scale para 1000m
resolucao_metros = 1000.
metros_por_grau = 111320 * np.cos(np.radians(lat_media))
grid_scale_graus = resolucao_metros / metros_por_grau

# monta os parametros
grid_params = helpers.fit_geometry(geometry=shapely_geometry,
                                   grid_crs='EPSG:4326',
                                   grid_scale=(grid_scale_graus, -grid_scale_graus))

# converte para Dataset
ds_AOD55 = xr.open_dataset(AOD55.mean(), engine = 'ee', **grid_params)

#----------------------------------------------------------------#
#                  MONÓXIDO DE CARBONO (CO)
#----------------------------------------------------------------#
# CO Sentinel-5P
S5P_co = ee.ImageCollection('COPERNICUS/S5P/NRTI/L3_CO') \
           .filterBounds(regiao_recorte) \
           .filterDate(datai,dataf)  \
           .select('CO_column_number_density')

# converte "ee.Geometry" para "shapely geometry" explicitamente
shapely_geometry = shape(regiao_recorte.getInfo())

# latitude central da geometria
centroide = shapely_geometry.centroid
lat_media = centroide.y  # em graus

# calcula o grid_scale para 1000m
resolucao_metros = 1000.
metros_por_grau = 111320 * np.cos(np.radians(lat_media))
grid_scale_graus = resolucao_metros / metros_por_grau

# monta os parametros
grid_params = helpers.fit_geometry(geometry=shapely_geometry,
                                   grid_crs='EPSG:4326',
                                   grid_scale=(grid_scale_graus, -grid_scale_graus))

# converte para Dataset
ds_S5P_co = xr.open_dataset(S5P_co.mean().multiply(2241.15), engine = 'ee', **grid_params)

#========================================================================================================================#
#                                               2) RGB e FALSA COR
#========================================================================================================================#
#----------------------------------------------------------------#
#                         GERA TIF
#----------------------------------------------------------------#
# parâmetros de visualização
param_vis_rgb = {'bands': ['B4','B3','B2'], 'min': 0.03, 'max': 0.19, 'gamma': 1.2}
param_vis_falsacor = {'bands': ['B12', 'B8A', 'B4'], 'min': 0.0, 'max': 0.3}

# imagem RGB PRÉ-FOGO
file_rgb_pre_fogo = f'{dir_output}/Parte_12_Fig_2a_rgb_pre_fogo.tif'
geemap.ee_export_image(imagem_pre.visualize(**param_vis_rgb),
                       filename=file_rgb_pre_fogo,
                       scale=20,
                       region=regiao_recorte,
                       file_per_band=False)

# imagem RGB PÓS-FOGO
file_rgb_pos_fogo = f'{dir_output}/Parte_12_Fig_2b_rgb_pos_fogo.tif'
geemap.ee_export_image(imagem_pos.visualize(**param_vis_rgb),
                       filename=file_rgb_pos_fogo,
                       scale=20,
                       region=regiao_recorte,
                       file_per_band=False)


# imagem FALSA COR PRÉ-FOGO
file_falsacor_pre_fogo = f'{dir_output}/Parte_12_Fig_2c_falsacor_pre_fogo.tif'
geemap.ee_export_image(imagem_pre.visualize(**param_vis_falsacor),
                       filename=file_falsacor_pre_fogo,
                       scale=20,
                       region=regiao_recorte,
                       file_per_band=False)

# imagem FALSA COR PÓS-FOGO
file_falsacor_pos_fogo = f'{dir_output}/Parte_12_Fig_2d_falsacor_pos_fogo.tif'
geemap.ee_export_image(imagem_pos.visualize(**param_vis_falsacor),
                       filename=file_falsacor_pos_fogo,
                       scale=20,
                       region=regiao_recorte,
                       file_per_band=False)

#----------------------------------------------------------------#
#                       LEITURA DO TIF
#----------------------------------------------------------------#
# função que faz a leitura da imagem TIF
def read_tif(file):

    # leitura da imagem TIF
    with rasterio.open(file) as src:

        # leitura das bandas
        b1 = src.read(1)
        b2 = src.read(2)
        b3 = src.read(3)

        # combinar as bandas em uma imagem. Normalizar os valores se necessário (dependendo do tipo de dados)
        if b1.dtype != np.uint8:

            # normalizar para 0-255 se for float ou outro tipo
            b1_norm = ((b1 - b1.min()) / (b1.max() - b1.min()) * 255).astype(np.uint8)
            b2_norm = ((b2 - b2.min()) / (b2.max() - b2.min()) * 255).astype(np.uint8)
            b3_norm = ((b3 - b3.min()) / (b3.max() - b3.min()) * 255).astype(np.uint8)
            combinado_image = np.dstack((b1_norm, b2_norm, b3_norm))

        else:

            combinado_image = np.dstack((b1, b2, b3))

        # obter as informações de georreferenciamento
        transform = src.transform
        bounds = src.bounds
        crs = src.crs

    return combinado_image, transform, bounds, crs

# leitura dos arquivos TIF
imagem_rgb_pre_fogo, transform, bounds, crs = read_tif(file_rgb_pre_fogo)
imagem_rgb_pos_fogo, transform, bounds, crs = read_tif(file_rgb_pos_fogo)
imagem_falsacor_pre_fogo, transform, bounds, crs = read_tif(file_falsacor_pre_fogo)
imagem_falsacor_pos_fogo, transform, bounds, crs = read_tif(file_falsacor_pos_fogo)

# extrai os limites da imagem
extent_rgb_e_falsacor = [bounds.left, bounds.right, bounds.bottom, bounds.top]

#========================================================================================================================#
#                                               3) NDVI
#========================================================================================================================#
#----------------------------------------------------------------#
#                         CÁLCULO DO NDVI
#----------------------------------------------------------------#
def ndvi(colecao):

    """
    Calcula o NDVI para uma coleção de imagens do Sentinel-2 mantendo todas as bandas originais

    Parâmetros:
    -----------
    colecao : ee.ImageCollection
        Coleção de imagens do Sentinel-2

    Retorna:
    --------
    ee.ImageCollection
        Coleção com todas as bandas originais + banda NDVI adicionada

    Função para calcular e adicionar NDVI a cada imagem
    """

    def adicionar_ndvi(imagem):

        # calcula NDVI e adiciona como nova banda à imagem.

        # NDVI = (NIR - Red) / (NIR + Red)
        # Sentinel-2: B8 = NIR, B4 = Red

        # calcula o índice NDVI
        ndvi = imagem.normalizedDifference(['B8', 'B4']).rename('NDVI')

        # adiciona a banda NDVI mantendo todas as bandas originais
        return imagem.addBands(ndvi)

    # aplica a função a todas as imagens da coleção
    return colecao.map(adicionar_ndvi)

# calcula o NDVI
sentinel2_com_ndvi = ndvi(sentinel2)

# seleciona as imagens com NDVI pré-fogo
imagem_pre_ndvi = sentinel2_com_ndvi.filterDate(data_pre_1, data_pre_2).mean().clip(regiao_recorte)

# seleciona as imagens com NDVI pós-fogo
imagem_pos_ndvi = sentinel2_com_ndvi.filterDate(data_pos_1, data_pos_2).mean().clip(regiao_recorte)

# NDVI pré-fogo
ndvi_pre = imagem_pre_ndvi.select('NDVI')

# NDVI pró-fogo
ndvi_pos = imagem_pos_ndvi.select('NDVI')

# calcula o dNDVI
dndvi = ndvi_pre.subtract(ndvi_pos).rename('dNDVI')

# classifica área de queimadas: limiar comum dndvi > 0.15
limiar_dndvi = 0.15
mascara_area_queimada_dndvi = dndvi.gt(limiar_dndvi)

#----------------------------------------------------------------#
#                     CÁLCULO DA ÁREA (KM²)
#----------------------------------------------------------------#
# cálcula a área de cada pixel em km²
calculo_area_queimada_km2_dndvi = mascara_area_queimada_dndvi.multiply(ee.Image.pixelArea()).divide(1000000)

# acumula todos os pixels em km², calculando assim a área total da queimada
soma_area_queimada_km2_dndvi = calculo_area_queimada_km2_dndvi.reduceRegion(reducer=ee.Reducer.sum(),
                                                                            geometry=regiao_recorte,
                                                                            scale=20, # resolução das bandas SWIR do Sentinel-2
                                                                            maxPixels=1e13,
                                                                            bestEffort=True)

# obtém valor em km²
valor_area_queimada_dndvi = soma_area_queimada_km2_dndvi.get('dNDVI').getInfo()

#----------------------------------------------------------------#
#          TRANSFORMA DE IMAGE PARA DATASET DO XARRAY
#----------------------------------------------------------------#
# converte "ee.Geometry" para "shapely geometry" explicitamente
shapely_geometry = shape(regiao_recorte.getInfo())

# latitude central da geometria
centroide = shapely_geometry.centroid
lat_media = centroide.y  # em graus

# calcula o grid_scale para 20m
resolucao_metros = 20.
metros_por_grau = 111320 * np.cos(np.radians(lat_media))
grid_scale_graus = resolucao_metros / metros_por_grau

# monta os parametros
grid_params = helpers.fit_geometry(geometry=shapely_geometry,
                                   grid_crs='EPSG:4326',
                                   grid_scale=(grid_scale_graus, -grid_scale_graus))

# converte para Dataset
ds_ndvi_pre = xr.open_dataset(ndvi_pre, engine = 'ee', **grid_params)
ds_ndvi_pos = xr.open_dataset(ndvi_pos, engine = 'ee', **grid_params)
ds_dndvi = xr.open_dataset(dndvi, engine = 'ee', **grid_params)
ds_dndvi_mascara_area_queimada = xr.open_dataset(mascara_area_queimada_dndvi.selfMask().rename('area_queimada'), engine = 'ee', **grid_params)

#========================================================================================================================#
#                                               4) NBR
#========================================================================================================================#
#----------------------------------------------------------------#
#                         CÁLCULO DO NBR
#----------------------------------------------------------------#
# função para calcular NBR (Sentinel-2: NIR=B8, SWIR=B12)
def nbr(image):
    return image.normalizedDifference(['B8', 'B12']).rename('NBR')

# NBR pré-fogo
nbr_pre = nbr(imagem_pre)

# NBR pós-fogo
nbr_pos = nbr(imagem_pos)

# calcula o dNBR
dnbr = nbr_pre.subtract(nbr_pos).rename('dNBR')

# classifica área de queimadas: limiar comum dNBR > 0.27
# valores acima de 0.27 geralmente indicam queimada MODERADA
limiar_dnbr = 0.27
mascara_area_queimada_dnbr = dnbr.gt(limiar_dnbr)

#----------------------------------------------------------------#
#                     CÁLCULO DA ÁREA (KM²)
#----------------------------------------------------------------#
# cálcula a área de cada pixel em km²
calculo_area_queimada_km2_dnbr = mascara_area_queimada_dnbr.multiply(ee.Image.pixelArea()).divide(1000000)

# acumula todos os pixels em km², calculando assim a área total da queimada
soma_area_queimada_km2_dnbr = calculo_area_queimada_km2_dnbr.reduceRegion(reducer=ee.Reducer.sum(),
                                                                          geometry=regiao_recorte,
                                                                          scale=20, # resolução das bandas SWIR do Sentinel-2
                                                                          maxPixels=1e13,
                                                                          bestEffort=True)

# obtém valor em km²
valor_area_queimada_dnbr = soma_area_queimada_km2_dnbr.get('dNBR').getInfo()

#----------------------------------------------------------------#
#          TRANSFORMA DE IMAGE PARA DATASET DO XARRAY
#----------------------------------------------------------------#
# converte "ee.Geometry" para "shapely geometry" explicitamente
shapely_geometry = shape(regiao_recorte.getInfo())

# latitude central da geometria
centroide = shapely_geometry.centroid
lat_media = centroide.y  # em graus

# calcula o grid_scale para 20m
resolucao_metros = 20.
metros_por_grau = 111320 * np.cos(np.radians(lat_media))
grid_scale_graus = resolucao_metros / metros_por_grau

# monta os parâmetros
grid_params = helpers.fit_geometry(geometry=shapely_geometry,
                                   grid_crs='EPSG:4326',
                                   grid_scale=(grid_scale_graus, -grid_scale_graus))

# converte para Dataset
ds_nbr_pre = xr.open_dataset(nbr_pre, engine = 'ee', **grid_params)
ds_nbr_pos = xr.open_dataset(nbr_pos, engine = 'ee', **grid_params)
ds_dnbr = xr.open_dataset(dnbr, engine = 'ee', **grid_params)
ds_dnbr_mascara_area_queimada = xr.open_dataset(mascara_area_queimada_dnbr.selfMask().rename('area_queimada'), engine = 'ee', **grid_params)

#========================================================================================================================#
#                                               5) ABAI
#========================================================================================================================#
#----------------------------------------------------------------#
#                         CÁLCULO DO ABAI
#----------------------------------------------------------------#
# função para calcular NBRSWIR (Sentinel-2: SWIR1=B11, SWIR2=B12)
def abai(image):
    return image.expression('(3 * SWIR2 - 2 * SWIR1 - 3 * GREEN) / (3 * SWIR2 + 2 * SWIR1 + 3 * GREEN)',
                            {'SWIR2': image.select('B12'),
                             'SWIR1': image.select('B11'),
                             'GREEN': image.select('B3')}).rename('ABAI')

# ABAI pré-fogo
abai_pre = abai(imagem_pre)

# ABAI pós-fogo
abai_pos = abai(imagem_pos)

# calcula o dABAI
dabai = abai_pre.subtract(abai_pos).rename('dABAI')

#----------------------------------------------------------------#
#               APLICA A MASCARA NA ÁREA QUEIMADA
#----------------------------------------------------------------#
# limiares
limiar_min, limiar_max = -1.0, -0.12

# seleciona regiões onde a condição é válida
mascara_area_queimada_dabai = dabai.gte(limiar_min).And(dabai.lte(limiar_max))

# aplica a mascara aos dados
area_queimada_mascarada_dabai = dabai.updateMask(mascara_area_queimada_dabai)

# coloca 1 nos pixels que obdecem o critério e 0 onde não obdece o critério
abai_binario = area_queimada_mascarada_dabai.multiply(0).add(1)

#----------------------------------------------------------------#
#                     CÁLCULO DA ÁREA (KM²)
#----------------------------------------------------------------#
# cálcula a área de cada pixel em km²:
calculo_area_queimada_km2_dabai = abai_binario.multiply(ee.Image.pixelArea()).divide(1000000)

# acumula todos os pixels em km², calculando assim a área total da queimada
soma_area_queimada_km2_dabai = calculo_area_queimada_km2_dabai.reduceRegion(reducer=ee.Reducer.sum(),
                                                                            geometry=regiao_recorte,
                                                                            scale=20, # resolução das bandas SWIR do Sentinel-2
                                                                            maxPixels=1e13,
                                                                            bestEffort=True)

# obtém valor em km²
valor_area_queimada_dabai = soma_area_queimada_km2_dabai.get('dABAI').getInfo()

#----------------------------------------------------------------#
#          TRANSFORMA DE IMAGE PARA DATASET DO XARRAY
#----------------------------------------------------------------#
# converte "ee.Geometry" para "shapely geometry" explicitamente
shapely_geometry = shape(regiao_recorte.getInfo())

# latitude central da geometria
centroide = shapely_geometry.centroid
lat_media = centroide.y  # em graus

# calcula o grid_scale para 20m
resolucao_metros = 20.
metros_por_grau = 111320 * np.cos(np.radians(lat_media))
grid_scale_graus = resolucao_metros / metros_por_grau

# monta os parâmetros
grid_params = helpers.fit_geometry(geometry=shapely_geometry,
                                   grid_crs='EPSG:4326',
                                   grid_scale=(grid_scale_graus, -grid_scale_graus))

# converte para Dataset
ds_abai_pre = xr.open_dataset(abai_pre, engine = 'ee', **grid_params)
ds_abai_pos = xr.open_dataset(abai_pos, engine = 'ee', **grid_params)
ds_dabai = xr.open_dataset(dabai, engine = 'ee', **grid_params)
ds_abai_mascara_area_queimada = xr.open_dataset(abai_binario.rename('area_queimada'), engine = 'ee', **grid_params)

#========================================================================================================================#
#                                               PLOTA FIGURAS
#========================================================================================================================#
# limites da imagem
lonmin, lonmax, latmin, latmax = -49.8169, -49.7098, -20.2338, -20.1358

# extrai os limites da imagem
extent = [lonmin, lonmax, latmin, latmax]

# cria a moldura da figura
fig, ax = uplt.subplots(axwidth=3, nrows=5, ncols=4, tight=True, proj='pcarree')

# define formato da figura
ax.format(coast=False, borders=False, innerborders=False,
          labels=True, latlines=0.05, lonlines=0.05,
          latlim=(latmin, latmax), lonlim=(lonmin, lonmax),
          small='15px', large='25px',
          linewidth=1.0, grid=True,
          suptitle='Queimadas - 24/agosto/2024\nDados: AQUA, NOAA-20 e Sentinel-2')

#--------------------------------------------------------------#
#             Linha 0: FOCOS DE CALOR E POLUIÇÃO
#--------------------------------------------------------------#
# FIRMS
map0 = ax[0].imshow(np.flipud(ds_firms['T21'][0,:,:]), cmap = 'inferno_r', extent = extent, levels = uplt.arange(30, 150, 10))
ax[0].set_extent(extent, crs=ccrs.PlateCarree())
ax[0].format(title='Temperatura: FIRMS', labels=[True, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[0].colorbar(map0, loc='b', label='°C', ticks=20, ticklabelsize=10, labelsize=10, width=0.2)

# VIIRS
map1 = ax[1].imshow(np.flipud(ds_ti4_viirs_noaa20['Bright_ti4'][0,:,:]), cmap = 'inferno_r', extent = extent, levels = uplt.arange(30, 150, 10))
ax[1].set_extent(extent, crs=ccrs.PlateCarree())
ax[1].format(title='Temperatura: VIIRS (NOAA-20)', labels=[False, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[1].colorbar(map1, loc='b', label='°C', ticks=20, ticklabelsize=10, labelsize=10, width=0.2)

# AOD055
map2 = ax[2].imshow(np.flipud(ds_AOD55['Optical_Depth_055'][0,:,:]), cmap = 'inferno_r', extent = extent, levels = uplt.arange(0, 0.8, 0.1))
ax[2].set_extent(extent, crs=ccrs.PlateCarree())
ax[2].format(title='AOD055', labels=[False, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[2].colorbar(map2, loc='b', label='', ticks=0.1, ticklabelsize=10, labelsize=10, width=0.2)

# MONÓXIDO DE CARBONO
map3 = ax[3].imshow(np.flipud(ds_S5P_co['CO_column_number_density'][0,:,:]), cmap = 'inferno_r', extent = extent, levels = uplt.arange(70, 140, 10))
ax[3].set_extent(extent, crs=ccrs.PlateCarree())
ax[3].format(title='Monóxido de Carbono (CO)', labels=[False, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[3].colorbar(map3, loc='b', label='DU', ticks=20, ticklabelsize=10, labelsize=10, width=0.2)

#--------------------------------------------------------------#
#                 Linha 1: RGB e FALSA COR
#--------------------------------------------------------------#
# leitura dos arquivos TIF
imagem_rgb_pre_fogo, transform, bounds, crs = read_tif(file_rgb_pre_fogo)
imagem_rgb_pos_fogo, transform, bounds, crs = read_tif(file_rgb_pos_fogo)
imagem_falsacor_pre_fogo, transform, bounds, crs = read_tif(file_falsacor_pre_fogo)
imagem_falsacor_pos_fogo, transform, bounds, crs = read_tif(file_falsacor_pos_fogo)

# extrai os limites da imagem
extent_rgb_e_falsacor = [bounds.left, bounds.right, bounds.bottom, bounds.top]

# RGB: pré-fogo
map4 = ax[4].imshow(imagem_rgb_pre_fogo, origin='upper', extent=extent, transform=ccrs.PlateCarree())
ax[4].set_extent(extent_rgb_e_falsacor, crs=ccrs.PlateCarree())
ax[4].format(title='RGB: Pré-fogo', labels=[True, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)

# RGB: pós-fogo
map5 = ax[5].imshow(imagem_rgb_pos_fogo, origin='upper', extent=extent, transform=ccrs.PlateCarree())
ax[5].set_extent(extent_rgb_e_falsacor, crs=ccrs.PlateCarree())
ax[5].format(title='RGB: Pós-fogo', labels=[False, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)

# FALSA COR: pré-fogo
map6 = ax[6].imshow(imagem_falsacor_pre_fogo, origin='upper', extent=extent, transform=ccrs.PlateCarree())
ax[6].set_extent(extent_rgb_e_falsacor, crs=ccrs.PlateCarree())
ax[6].format(title='FALSA COR: Pré-fogo', labels=[False, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)

# FALSA COR: pós-fogo
map7 = ax[7].imshow(imagem_falsacor_pos_fogo, origin='upper', extent=extent, transform=ccrs.PlateCarree())
ax[7].set_extent(extent_rgb_e_falsacor, crs=ccrs.PlateCarree())
ax[7].format(title='FALSA COR: Pós-fogo', labels=[False, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)

#"""
#--------------------------------------------------------------#
#                      Linha 2: NDVI
#--------------------------------------------------------------#
# pré-fogo
map8 = ax[8].imshow(np.flipud(ds_ndvi_pre['NDVI'][0,:,:]), cmap = 'RdYlGn', extent = extent, levels = uplt.arange(-0.6, 0.6, 0.1))
ax[8].set_extent(extent, crs=ccrs.PlateCarree())
ax[8].format(title='NDVI: Pré-fogo', labels=[True, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[8].colorbar(map8, loc='b', label='NDVI', ticks=0.2, ticklabelsize=10, labelsize=10, width=0.2)

# pós-fogo
map9 = ax[9].imshow(np.flipud(ds_ndvi_pos['NDVI'][0,:,:]), cmap = 'RdYlGn', extent = extent, levels = uplt.arange(-0.6, 0.6, 0.1))
ax[9].set_extent(extent, crs=ccrs.PlateCarree())
ax[9].format(title='NDVI: Pós-fogo', labels=[False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[9].colorbar(map9, loc='b', label='NDVI', ticks=0.2, ticklabelsize=10, labelsize=10, width=0.2)


# dNDVI
map10 = ax[10].imshow(np.flipud(ds_dndvi['dNDVI'][0,:,:]), cmap = 'coolwarm', extent = extent, levels = uplt.arange(-0.20, 0.20, 0.05))
ax[10].set_extent(extent, crs=ccrs.PlateCarree())
ax[10].format(title='dNDVI: Pré - Pós', labels=[False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[10].colorbar(map10, loc='b', label='dNDVI', ticks=0.05, ticklabelsize=10, labelsize=10, width=0.2)

# área queimada
map11 = ax[11].imshow(np.flipud(ds_dndvi_mascara_area_queimada['area_queimada'][0,:,:]), cmap = 'RdYlGn', extent = extent)
ax[11].set_extent(extent, crs=ccrs.PlateCarree())
ax[11].format(title=f'Área Queimada: dNDVI > {limiar_dndvi}', labels=[False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
stats_text = (f'Período Pré-Fogo: {data_pre_1}\n'
              f'Período Pós-Fogo: {data_pos_1}\n'
              f'Área Queimada: {valor_area_queimada_dndvi:.1f} km²')
ax[11].text(0.03, 0.02, stats_text,
            transform=ax[11].transAxes,
            color='black',
            fontsize=9,
            verticalalignment='bottom',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.9))

#--------------------------------------------------------------#
#                      Linha 3: NBR
#--------------------------------------------------------------#
# pré-fogo
map12 = ax[12].imshow(np.flipud(ds_nbr_pre['NBR'][0,:,:]), cmap = 'RdYlGn', extent = extent, levels = uplt.arange(-0.35, 0.35, 0.1))
ax[12].set_extent(extent, crs=ccrs.PlateCarree())
ax[12].format(title='NBR: Pré-fogo', labels=[True, False, False, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[12].colorbar(map12, loc='b', label='NBR', ticks=0.1, ticklabelsize=10, labelsize=10, width=0.2)

# pós-fogo
map13 = ax[13].imshow(np.flipud(ds_nbr_pos['NBR'][0,:,:]), cmap = 'RdYlGn', extent = extent, levels = uplt.arange(-0.35, 0.35, 0.1))
ax[13].set_extent(extent, crs=ccrs.PlateCarree())
ax[13].format(title='NBR: Pós-fogo', labels=[False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[13].colorbar(map13, loc='b', label='NBR', ticks=0.1, ticklabelsize=10, labelsize=10, width=0.2)

# dNBR
map14 = ax[14].imshow(np.flipud(ds_dnbr['dNBR'][0,:,:]), cmap = 'coolwarm', extent = extent, levels = uplt.arange(-0.5, 0.5, 0.1))
ax[14].set_extent(extent, crs=ccrs.PlateCarree())
ax[14].format(title='dNBR: Pré - Pós', labels=[False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[14].colorbar(map14, loc='b', label='dNBR', ticks=0.1, ticklabelsize=10, labelsize=10, width=0.2)

# área queimada
map15 = ax[15].imshow(np.flipud(ds_dnbr_mascara_area_queimada['area_queimada'][0,:,:]), cmap = 'RdYlGn', extent = extent)
ax[15].set_extent(extent, crs=ccrs.PlateCarree())
ax[15].format(title=f'Área Queimada: dNBR > {limiar_dnbr}', labels=[False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
stats_text = (f'Período Pré-Fogo: {data_pre_1}\n'
              f'Período Pós-Fogo: {data_pos_1}\n'
              f'Área Queimada: {valor_area_queimada_dnbr:.1f} km²')
ax[15].text(0.03, 0.02, stats_text,
            transform=ax[15].transAxes,
            color='black',
            fontsize=9,
            verticalalignment='bottom',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.9))

#--------------------------------------------------------------#
#                      Linha 4: ABAI
#--------------------------------------------------------------#
# pré-fogo
map16 = ax[16].imshow(np.flipud(ds_abai_pre['ABAI'][0,:,:]), cmap = 'RdYlGn_r', extent = extent, levels = uplt.arange(-0.4, 0.1, 0.1))
ax[16].set_extent(extent, crs=ccrs.PlateCarree())
ax[16].format(title='ABAI: Pré-fogo', labels=[True, False, True, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[16].colorbar(map16, loc='b', label='ABAI', ticks=0.1, ticklabelsize=10, labelsize=10, width=0.2, space=1.3)

# pós-fogo
map17 = ax[17].imshow(np.flipud(ds_abai_pos['ABAI'][0,:,:]), cmap = 'RdYlGn_r', extent = extent, levels = uplt.arange(-0.4, 0.1, 0.1))
ax[17].set_extent(extent, crs=ccrs.PlateCarree())
ax[17].format(title='ABAI: Pós-fogo', labels=[False, False, True, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[17].colorbar(map17, loc='b', label='ABAI', ticks=0.1, ticklabelsize=10, labelsize=10, width=0.2, space=0.3)

# dABAI
map18 = ax[18].imshow(np.flipud(ds_dabai['dABAI'][0,:,:]), cmap = 'coolwarm', extent = extent, levels = uplt.arange(-0.3, 0.3, 0.1))
ax[18].set_extent(extent, crs=ccrs.PlateCarree())
ax[18].format(title='dABAI: Pré - Pós', labels=[False, False, True, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
ax[18].colorbar(map18, loc='b', label='dABAI', ticks=0.1, ticklabelsize=10, labelsize=10, width=0.2, space=0.3)

# área queimada
map19 = ax[19].imshow(np.flipud(ds_abai_mascara_area_queimada['area_queimada'][0,:,:]), cmap = 'RdYlGn', extent = extent)
ax[19].set_extent(extent, crs=ccrs.PlateCarree())
ax[19].format(title=f'Área Queimada: dABAI entre {limiar_min} e {limiar_max}', labels=[False, False, True, False], linewidth=1, titleloc='c', titlecolor='black', titlesize=10)
stats_text = (f'Período Pré-Fogo: {data_pre_1}\n'
              f'Período Pós-Fogo: {data_pos_1}\n'
              f'Área Queimada: {valor_area_queimada_dabai:.1f} km²')
ax[19].text(0.03, 0.02, stats_text,
            transform=ax[19].transAxes,
            color='black',
            fontsize=9,
            verticalalignment='bottom',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.9))

# salva figura
fig.savefig(f'{dir_output}/Parte_12e_analises_principais.jpg', dpi=300, bbox_inches="tight")

# mostra figura na tela
plt.show()
```
