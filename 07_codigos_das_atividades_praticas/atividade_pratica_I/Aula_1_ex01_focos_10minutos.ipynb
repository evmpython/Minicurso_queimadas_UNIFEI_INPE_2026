```python
# DataFrame é um formato tabular de dados (linhas x colunas igual ao Excel).

# Leitura do arquivo específico.
# https://dataserver-coids.inpe.br/queimadas/queimadas/focos/csv/10min/
df = pd.read_csv("https://dataserver-coids.inpe.br/queimadas/queimadas/focos/csv/10min/focos_10min_20260727_1440.csv")

# Visualiza o DataFrame.
df
```

## Plotagem estática dos focos de queimadas

- **Pergunta 2.**

```python
# Converte os pontos do df (DataFrame) para geometry, isso gera o GeoDataFrame.
# lon e lat são as colunas do df.
gdf_focos = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df.lon, df.lat), crs="epsg:4326")

# Visualização do gdf_focos (GeoDataFrame).
gdf_focos
```

```python
# # Checa o tipo do dado apos converter para GeoDataFrame.
gdf_focos.info()
```

```python
# Plot básico do gdf (GeoDataFrame).
gdf_focos.plot()
```

### Leitura do GeoDataFrame

```python
# URL do shapefile global.
url = "https://naturalearth.s3.amazonaws.com/10m_cultural/ne_10m_admin_0_countries.zip"

# Leitura do shapefile.
gdf_global = gpd.read_file(url)

# Visualização do gdf_global.
gdf_global
```

```python
# Retorna o número de linhas e de colunas do GeoDataFrame.
gdf_global.shape
```

```python
# Nome dos continentes.
# O value_counts() retorna os valores que não se repetem.
gdf_global["CONTINENT"].value_counts()
```

```python
# O operador "==" é usado como operador de igualdade lógica. É usado para comparar igualdade.
# Compara o que está do lado esquerdo com o que está do lado direito, se foram iguais,
# recebe True, caso contrário, False.

# Seleciona somente América do Sul.
filtro = gdf_global["CONTINENT"] == "South America" # Retorna um booleano (True ou False).

# Aplica o filtro.
# gdf_america_sul = GeoDataFrame da América do Sul.
gdf_america_sul = gdf_global[filtro]

# Plot básico do gdf.
gdf_america_sul.plot()
```

### Verificação do sistema de coordenada espacial

- **Pergunta 3.**

```python
# Checar o Coordinate Reference System (CRS) do gdf.
# Aqui usa-se o arquivo de focos de queimadas a cada 10 minutos.
gdf_focos.crs
```

```python
# Checar o Coordinate Reference System (CRS) do arquivo shapefile aberto.
gdf_america_sul.crs
```

### Interseção entre dados de focos e o shapefile

```python
# Realiza a interseção espacial para mascarar os dados.
# O gdf_as é considerado uma máscara. Os focos de queimadas são
# considerados apenas dentro do domínio do gdf_as.
focos_mascarado = gpd.overlay(gdf_focos, gdf_america_sul, how="intersection")

# Visualiza o GeoDataFrame.
focos_mascarado
```

### Plot do mapa estático dos focos de queimadas

```python
# Plot dos focos de queimadas a cada 10 minutos sobre o mapa.

# Define o tamanho da imagem.
fig, ax = plt.subplots(figsize=(10, 10))

# Plot do gdf_america_sul.
gdf_america_sul.plot(ax=ax, color="lightgray", edgecolor="black", legend=True)

# Plot dos focos de queimadas.
focos_mascarado.plot(ax=ax, marker="o", color="red", markersize=5, label="Focos de Queimadas")

# Formatação da figura.
ax.set_title("Focos de Queimada na América do Sul")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.legend()
plt.grid(True, linestyle="--", alpha=0.7)

# Salva a figura.
plt.savefig(f"{diretorio_figuras}/ex01-01focos_america_sul.png", dpi=300)

# Mostra a figura.
plt.show()

# Libera memória.
plt.close()
```

## Plotagem dinâmica dos focos de queimadas usando leafmap

O leafmap é um pacote Python para mapeamento interativo e análise geoespacial com o mínimo de código em um ambiente Jupyter.

Mais informações, acessar: https://leafmap.org

### Plot inicial do leafmap

```python
# Importação da biblioteca leafmap.
import leafmap

# Define o ponto central do mapa e nível de zoom.
# Quanto menor o nível de zoom, mais amplo será o mapa.
# zoom = 0 a 18.
# -15: latitude e -60: longitude.
Map = leafmap.Map(center=[-15, -60], zoom=4)

# adiciona mapa de fundo.
Map.add_basemap("SATELLITE")

# Mostra o mapa.
Map
```

```python
# Apenas mostrando novamente o DataFrame "df" criado no início do script para
# ver o nome das colunas que serão utilizadas para criar o mapa interativo.
df
```
