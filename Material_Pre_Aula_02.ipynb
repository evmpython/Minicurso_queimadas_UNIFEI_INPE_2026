Agora para acessar os dados do GEE precisamos autenticar e inicializar o GEE. Este procedimento irá nos permitir ter acesso ao banco de dados e a todas as funcionalidades do GEE.
Para isto usaremos os seguintes comandos:

![Texto alternativo](https://github.com/evmpython/Minicurso_queimadas_UNIFEI_INPE_2026/blob/main/04_logos/autenticando_gee.png?raw=true)

2. `ee.Initialize(project='SUA ID PROJECT', opt_url='https://earthengine-highvolume.googleapis.com')`: você deverá informar sua "ID project". O atributo *opt_url*  específica o endpoint da API do GEE, usa-se um servidor de alto volume (highvolume) para processamento pesado.

```python
ee.Initialize(project='SUA ID PROJECT', opt_url='https://earthengine-highvolume.googleapis.com')
```

Se deu tudo certo, significa que agora você esta com acesso liberado ao GEE. Para gerar mapas no GEE iremos utilizar a biblioteca [geemap](https://geemap.org/). O geemap estende o GEE com ferramentas de visualização interativa, especialmente útil em notebooks Jupyter/Colab. Assim oremos importar a biblioteca geemap.

```python
# importando a biblioteca geemap
import geemap
```

Agora vamos criar nosso mapa com  geemap. Neste exemplo estamos plotando composição RGB do sensor MODIS.

```python
# carrega os dados
modis = ee.ImageCollection('MODIS/061/MYD09CMG').filterDate('2024-08-05','2024-08-06')

# cria a moldura do mapa
Map = geemap.Map()

# parâmetros de visualização
param_vis = {'bands': ['Coarse_Resolution_Surface_Reflectance_Band_1',
                       'Coarse_Resolution_Surface_Reflectance_Band_4',
                       'Coarse_Resolution_Surface_Reflectance_Band_3'],
             'min': -0.01, 'max': 1.6}

# adicionando basemap
Map.add_basemap('SATELLITE')

# plotando mapa
Map.addLayer(modis, param_vis, 'RGB AQUA')

# exibe na tela
Map
```
